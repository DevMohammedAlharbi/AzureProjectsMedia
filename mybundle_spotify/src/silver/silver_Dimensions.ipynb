### DimArtist

```python
# READ must be before WRITE 
# Read it to put it in " df_art "  ==  then i can transormatin == when write in new path 
# Why ? Because write saves df_art , and df_art is only created by the read cell . Running write first throws an error: df_art is not defined !


df_art = spark.readStream.format("cloudFiles")\
         .option("cloudFiles.format", "parquet")\
         .option("cloudFiles.schemaLocation" , "abfss://silver@storge4azureproject.dfs.core.windows.net/DimArtist/checkpoint")\
         .option("schemaEvolutionMode", "addNewColumns")\
         .load("abfss://bronze@storge4azureproject.dfs.core.windows.net/DimArtist") 
         
```

```python
display(df_art, checkpointLocation="abfss://silver@storge4azureproject.dfs.core.windows.net/DimArtist/display_checkpoint")
```

```python
display(spark.read.format("parquet").load("abfss://bronze@storge4azureproject.dfs.core.windows.net/DimArtist"))
```

```python
path = "abfss://silver@storge4azureproject.dfs.core.windows.net/DimArtist/display_checkpoint"
dbutils.fs.rm(path, True)
display(df_art, checkpointLocation=path)
```

```python
df_art_obj = reusable()
df_art = df_art_obj.dropColumns(df_art, ['_rescued_data'])
df_art = df_art.dropDuplicates(['artist_id'])
```

```python
# Then write ! 
df_art.writeStream.format("delta")\
    .outputMode("append")\
    .option("checkpointLocation", "abfss://silver@storge4azureproject.dfs.core.windows.net/DimArtist/checkpoint")\
    .trigger(availableNow=True)\
    .option("path","abfss://silver@storge4azureproject.dfs.core.windows.net/DimArtist/data")\
    .toTable("spotify_cata.silver.DimArtist") # لو بحفظ الملفات واسوي جدول بنفس الوقت

```

### DimTrack

READ THEN TRANSOR THEN WRITE 

```python
df_track = spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format", "parquet")\
    .option("cloudFiles.schemaLocation" , "abfss://silver@storge4azureproject.dfs.core.windows.net/DimTrack/checkpoint")\
    .option("schemaEvolutionMode", "addNewColumns")\
    .load("abfss://bronze@storge4azureproject.dfs.core.windows.net/DimTrack")
```

```python
df_track = df_track.withColumn( "duration_flag" , when( col("duration_sec")<150 , "short" )\
                                                  .when( col("duration_sec")<300 , "medium" )\
                                                  .otherwise("long") )
```

```python
df_track.printSchema()   # duration_flag should appear at the end
```

```python
for q in spark.streams.active:
    q.stop()
```

```python
# FUN REPLAYC -- > regexp_replace( col("col name") , "what i want to replace", "what i want" )

df_track = df_track.withColumn( "track_name" , regexp_replace( col("track_name") , "-", " ") ) 
```

```python
df_track_obj = reusable()
df_track = df_track_obj.dropColumns(df_track, ['_rescued_data'])
```

```python
df_track.printSchema() 
```

```python
df_track.writeStream.format("delta")\
    .outputMode("append")\
    .option("checkpointLocation", "abfss://silver@storge4azureproject.dfs.core.windows.net/DimTrack/checkpoint")\
    .trigger(availableNow=True)\
    .option("path","abfss://silver@storge4azureproject.dfs.core.windows.net/DimTrack/data")\
    .toTable("spotify_cata.silver.DimTrack") # لو بحفظ الملفات واسوي جدول بنفس الوقت
```
